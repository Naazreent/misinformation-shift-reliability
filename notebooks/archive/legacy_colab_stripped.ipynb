```
# Function to remove outliers using IQR
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    # Define outlier boundaries
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    # Remove outliers
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Remove outliers in 'text_length' and 'word_count'
data = remove_outliers(data, 'text_length')
data = remove_outliers(data, 'word_count')

# Verify the shape of the dataset
print(f"Data shape after outlier removal: {data.shape}")
```

```
from tqdm import tqdm

def train(model, train_loader, val_loader, optimizer, criterion, epochs=3):
    model.train()
    for epoch in range(epochs):
        total_loss = 0
        print(f"Epoch {epoch + 1}/{epochs}")
        for batch in tqdm(train_loader):
            input_ids, attention_mask, labels = [b.to(device) for b in batch]

            optimizer.zero_grad()

            # Forward pass
            outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
            loss = outputs.loss
            total_loss += loss.item()

            # Backward pass
            loss.backward()
            optimizer.step()

        avg_loss = total_loss / len(train_loader)
        print(f"Epoch {epoch + 1}/{epochs}, Loss: {avg_loss:.4f}")

        # Validation step
        validate(model, val_loader, criterion)

def validate(model, val_loader, criterion):
    model.eval()
    total_val_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():
        for batch in val_loader:
            input_ids, attention_mask, labels = [b.to(device) for b in batch]

            # Forward pass
            outputs = model(input_ids, attention_mask=attention_mask)
            loss = criterion(outputs.logits, labels)
            total_val_loss += loss.item()

            preds = torch.argmax(outputs.logits, dim=1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)

    avg_val_loss = total_val_loss / len(val_loader)
    accuracy = correct / total
    print(f"Validation Loss: {avg_val_loss:.4f}, Accuracy: {accuracy:.4%}")
```

```
import torch
import torch.nn as nn

class CNNClassifier(nn.Module):
    def __init__(self, vocab_size, embed_size, num_classes):
        super(CNNClassifier, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_size)
        self.conv1 = nn.Conv2d(1, 100, (3, embed_size))  # 100 filters of size 3
        self.conv2 = nn.Conv2d(1, 100, (4, embed_size))  # 100 filters of size 4
        self.conv3 = nn.Conv2d(1, 100, (5, embed_size))  # 100 filters of size 5
        self.dropout = nn.Dropout(0.5)
        self.fc = nn.Linear(300, num_classes)  # 300 = 100 * 3 filters

    def forward(self, x):
        x = self.embedding(x)  # Shape: (batch_size, seq_length, embed_size)
        x = x.unsqueeze(1)  # Add a channel dimension: (batch_size, 1, seq_length, embed_size)

        # Apply convolution and activation
        x1 = torch.relu(self.conv1(x)).squeeze(3)  # Remove last dimension
        x2 = torch.relu(self.conv2(x)).squeeze(3)
        x3 = torch.relu(self.conv3(x)).squeeze(3)

        # Max pooling
        x1 = torch.max_pool1d(x1, x1.size(2)).squeeze(2)
        x2 = torch.max_pool1d(x2, x2.size(2)).squeeze(2)
        x3 = torch.max_pool1d(x3, x3.size(2)).squeeze(2)

        # Concatenate pooled features
        x = torch.cat((x1, x2, x3), dim=1)
        x = self.dropout(x)
        x = self.fc(x)  # Fully connected layer
        return x
```

```
vocab_size = tokenizer.vocab_size  # Vocabulary size from the tokenizer
embed_size = 128  # Embedding size
num_classes = 2  # Binary classification: Fake or True

cnn_model = CNNClassifier(vocab_size, embed_size, num_classes)
cnn_model = cnn_model.to(device)

# Loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(cnn_model.parameters(), lr=1e-4)
```

```
import matplotlib.pyplot as plt
import networkx as nx

def draw_fnn():
    G = nx.DiGraph()

    # Define layers
    input_nodes = ["Input " + str(i) for i in range(1, 4)]
    hidden_nodes = ["Hidden " + str(i) for i in range(1, 6)]
    output_nodes = ["Output"]

    # Add edges
    for input_node in input_nodes:
        for hidden_node in hidden_nodes:
            G.add_edge(input_node, hidden_node)

    for hidden_node in hidden_nodes:
        for output_node in output_nodes:
            G.add_edge(hidden_node, output_node)

    # Draw graph
    pos = nx.spring_layout(G, seed=42)
    nx.draw(G, pos, with_labels=True, node_size=2000, node_color="lightblue", font_size=10, font_weight="bold")
    plt.title("Feedforward Neural Network (FNN)", fontsize=16)
    plt.show()

def draw_cnn():
    plt.figure(figsize=(10, 6))

    # Convolutional layer
    plt.barh(4, 2, left=0, color="lightblue", label="Input Image")
    plt.barh(3, 2, left=2, color="orange", label="Convolution Layer")
    plt.barh(2, 1, left=4, color="green", label="Pooling Layer")
    plt.barh(1, 0.5, left=5, color="red", label="Fully Connected Layer")

    # Labels and title
    plt.yticks([1, 2, 3, 4], ["Fully Connected", "Pooling", "Convolution", "Input"])
    plt.xlabel("Processing Steps", fontsize=12)
    plt.title("Convolutional Neural Network (CNN)", fontsize=16)
    plt.legend()
    plt.show()

def draw_lstm():
    G = nx.DiGraph()

    # Define layers
    input_nodes = ["Input Sequence"]
    lstm_nodes = ["LSTM Cell " + str(i) for i in range(1, 4)]
    output_nodes = ["Output"]

    # Add edges
    for i in range(len(lstm_nodes) - 1):
        G.add_edge(lstm_nodes[i], lstm_nodes[i + 1])

    G.add_edge(input_nodes[0], lstm_nodes[0])
    G.add_edge(lstm_nodes[-1], output_nodes[0])

    # Draw graph
    pos = nx.spring_layout(G, seed=42)
    nx.draw(G, pos, with_labels=True, node_size=2000, node_color="lightgreen", font_size=10, font_weight="bold")
    plt.title("Long Short-Term Memory (LSTM)", fontsize=16)
    plt.show()

def draw_bert():
    plt.figure(figsize=(12, 8))

    # BERT layers illustration
    plt.barh(6, 2, left=0, color="lightblue", label="Input Embedding")
    plt.barh(5, 2, left=2, color="orange", label="Transformer Encoder 1")
    plt.barh(4, 2, left=4, color="orange", label="Transformer Encoder 2")
    plt.barh(3, 2, left=6, color="orange", label="Transformer Encoder N")
    plt.barh(2, 1, left=8, color="green", label="Attention Mechanism")
    plt.barh(1, 0.5, left=9, color="red", label="Prediction Layer")

    # Labels and title
    plt.yticks([1, 2, 3, 4, 5, 6], [
        "Prediction Layer", "Attention", "Transformer N",
        "Transformer 2", "Transformer 1", "Input Embedding"
    ])
    plt.xlabel("Processing Steps", fontsize=12)
    plt.title("Bidirectional Encoder Representations from Transformers (BERT)", fontsize=16)
    plt.legend()
    plt.show()

# Draw all models
draw_fnn()
draw_cnn()
draw_lstm()
draw_bert()
```

```
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Data from the classification report
data = {
    "Class": ["Fake", "True"],
    "Precision": [0.95, 0.96],
    "Recall": [0.89, 0.98],
    "F1-Score": [0.92, 0.97]
}

# Convert to a DataFrame
df = pd.DataFrame(data)

# Bar Chart
plt.figure(figsize=(10, 5))
x = np.arange(len(df["Class"]))  # the label locations
width = 0.2  # the width of the bars

fig, ax = plt.subplots(figsize=(8, 5))
rects1 = ax.bar(x - width, df["Precision"], width, label='Precision', color='skyblue')
rects2 = ax.bar(x, df["Recall"], width, label='Recall', color='orange')
rects3 = ax.bar(x + width, df["F1-Score"], width, label='F1-Score', color='green')

# Add labels, title, and legend
ax.set_xlabel('Class')
ax.set_ylabel('Scores')
ax.set_title('Classification Report Metrics by Class')
ax.set_xticks(x)
ax.set_xticklabels(df["Class"])
ax.legend()

# Show plot
plt.tight_layout()
plt.show()

# Heatmap
heatmap_data = df.set_index("Class").T  # Transpose for heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(heatmap_data, annot=True, cmap="Blues", cbar=True)
plt.title("Classification Report Heatmap")
plt.show()
```
